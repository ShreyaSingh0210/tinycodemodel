```python
import torch

# deliberately wrong weights - negative, so model predicts opposite
weights = torch.tensor([-0.5, -0.3], requires_grad=True)
bias = torch.tensor(0.0, requires_grad=True)
student = torch.tensor([5.0, 7.0])
actual = torch.tensor(1.0)  # student passed

for step in range(100):
    output = (student * weights).sum() + bias
    prediction = torch.sigmoid(output)
    loss = (prediction - actual) ** 2

    loss.backward()

    with torch.no_grad():
        weights -= 0.1 * weights.grad
        bias -= 0.1 * bias.grad
        weights.grad.zero_()
        bias.grad.zero_()

    if step % 10 == 0:
        print(f"Step {step:3d} | Loss: {loss.item():.4f} | Prediction: {prediction.item():.4f}")
```

```python
import urllib.request

# download real Python code from GitHub
url = "https://raw.githubusercontent.com/karpathy/char-rnn/master/data/tinyshakespeare/input.txt"
urllib.request.urlretrieve(url, "data.txt")

with open("data.txt", "r") as f:
    text = f.read()

print(f"Dataset size: {len(text):,} characters")
```

```python
import torch
import torch.nn as nn

# build vocabulary from new data
chars = sorted(set(text))
vocab_size = len(chars)
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for i, ch in enumerate(chars)}

print(f"Vocab size: {vocab_size}")
print(f"Data size: {len(text):,}")

# encode text
data = torch.tensor([stoi[ch] for ch in text], dtype=torch.long)

# context window - model sees last 32 characters
block_size = 32

# create batches properly
def get_batch(batch_size=64):
    ix = torch.randint(len(data) - block_size, (batch_size,))
    x = torch.stack([data[i:i+block_size] for i in ix])
    y = torch.stack([data[i+1:i+block_size+1] for i in ix])
    return x, y

# test it
xb, yb = get_batch()
print(f"Input batch shape:  {xb.shape}")
print(f"Target batch shape: {yb.shape}")
```

```python
class BetterCodeModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=64, hidden_dim=256, block_size=32):
        super().__init__()
        self.block_size = block_size
        self.embed = nn.Embedding(vocab_size, embed_dim)
        self.net = nn.Sequential(
            nn.Linear(embed_dim * block_size, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, vocab_size)
        )

    def forward(self, x):
        embedded = self.embed(x)
        flattened = embedded.view(x.shape[0], -1)
        return self.net(flattened)

model = BetterCodeModel(vocab_size)
total = sum(p.numel() for p in model.parameters())
print(f"Parameters: {total:,}")

# train
optimizer = torch.optim.Adam(model.parameters(), lr=3e-4)
loss_fn = nn.CrossEntropyLoss()

print("Training...")
for step in range(5000):
    xb, yb = get_batch()
    logits = model(xb)
    loss = loss_fn(logits, yb[:, -1])
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if step % 1000 == 0:
        print(f"Step {step:5d} | Loss: {loss.item():.4f}")
```

```python
def generate(model, start_text, max_chars=300):
    model.eval()
    with torch.no_grad():
        # encode starting text
        idx = torch.tensor(
            [stoi[c] for c in start_text[-model.block_size:]]
        ).unsqueeze(0)

        # pad if needed
        if idx.shape[1] < model.block_size:
            pad = torch.zeros(
                1, model.block_size - idx.shape[1], dtype=torch.long
            )
            idx = torch.cat([pad, idx], dim=1)

        result = start_text
        for _ in range(max_chars):
            logits = model(idx)
            probs = torch.softmax(logits[0], dim=-1)
            next_idx = torch.multinomial(probs, 1)
            result += itos[next_idx.item()]
            # slide the window
            idx = torch.cat([
                idx[:, 1:],
                next_idx.unsqueeze(0)
            ], dim=1)

    return result

print("=== Generated Text ===")
print(generate(model, "def "))
print("\n=== Generated Text ===")
print(generate(model, "for i"))
```
